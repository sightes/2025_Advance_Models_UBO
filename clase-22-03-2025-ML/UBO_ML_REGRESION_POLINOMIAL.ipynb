```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
# 1. CREACIÓN DEL DATASET (velocidad vs distancia de frenado)
data = {
    "Velocidad": [
        4,  4,  7,  7,  8,  9, 10, 10, 10, 11,
        11, 12, 12, 12, 13, 13, 13, 14, 14, 14,
        15, 15, 16, 16, 17, 17, 17, 18, 18, 18,
        19, 19, 19, 20, 20, 20, 20, 22, 23, 24,
        24, 24, 25, 25, 26, 26, 27, 27, 28, 28
    ],
    "Distancia": [
         2, 10,  4, 22, 16, 10, 18, 26, 34, 17,
        28, 14, 20, 24, 28, 26, 34, 26, 36, 60,
        20, 26, 32, 40, 32, 40, 50, 42, 56, 76,
        36, 46, 68, 32, 48, 52, 56, 66, 54, 70,
        92, 93, 85, 63, 80, 90, 97, 95, 98, 100
    ]
}

df = pd.DataFrame(data)
```

```python
# X son las velocidades (variable independiente)
X = df["Velocidad"].values.reshape(-1, 1)
```

```python
# y es la distancia de frenado (variable dependiente)
y = df["Distancia"].values
```

```python
# 2. GRÁFICO INICIAL DE LOS PUNTOS DEL DATASET
plt.scatter(X, y)
plt.title("Datos reales: Distancia de frenado vs Velocidad")
plt.xlabel("Velocidad")
plt.ylabel("Distancia de frenado")
plt.show()
```

```python
# 3. CREACIÓN Y ENTRENAMIENTO DEL MODELO DE REGRESIÓN POLINOMIAL DE GRADO 3
poly_features = PolynomialFeatures(degree=3)
X_poly = poly_features.fit_transform(X)
```

```python
model = LinearRegression()
model.fit(X_poly, y)
```

```python
# 4. PREDICCIÓN PARA GRAFICAR LA CURVA POLINOMIAL
X_nuevo = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
X_nuevo_poly = poly_features.transform(X_nuevo)
y_nuevo = model.predict(X_nuevo_poly)
```

```python
# 5. GRÁFICO FINAL: PUNTOS ORIGINALES + CURVA DE LA REGRESIÓN POLINOMIAL
plt.scatter(X, y)
plt.plot(X_nuevo, y_nuevo, linewidth=2)  # Línea de la regresión polinomial
plt.title("Regresión Polinomial (Grado 3)")
plt.xlabel("Velocidad")
plt.ylabel("Distancia de frenado")
plt.show()
```

```python
# Impresión de los coeficientes y el intercepto del modelo:
print("Coeficientes (a1, a2, a3):", model.coef_)
print("Intercepto (b):", model.intercept_)
```
