```python
# ========================================================
# Ejemplo de Regresión Lineal con dataset sintético
# ========================================================
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# ---------------------------------
# 1. Generar datos sintéticos
# ---------------------------------
np.random.seed(42)  # Para reproducibilidad
num_muestras = 100
```

```python
# Metros cuadrados: valores entre 50 y 200, distribuidos aleatoriamente
X = 50 + (200 - 50) * np.random.rand(num_muestras, 1)
```

```python
# Supongamos que el precio base es 20,000 y cada metro cuadrado agrega 300
# Se añade un ruido aleatorio a los precios
y = 20000 + 300 * X + np.random.randn(num_muestras, 1) * 5000
```

```python
# ---------------------------------
# 2. Crear modelo de Regresión
# ---------------------------------
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
# ---------------------------------
# 3. Mostrar datos (gráfico de puntos)
# ---------------------------------
plt.scatter(X, y)
plt.title("Puntos del Dataset (Metros cuadrados vs. Precio)")
plt.xlabel("Metros cuadrados")
plt.ylabel("Precio de la casa")
plt.show()
```

```python
# ---------------------------------
# 4. Graficar la recta de regresión
# ---------------------------------
# Generar valores para la predicción
X_nuevo = np.linspace(50, 200, 100).reshape(-1, 1)
y_pred = modelo.predict(X_nuevo)
plt.scatter(X, y, label="Datos reales")
plt.plot(X_nuevo, y_pred, label="Línea de regresión")
plt.title("Recta de Regresión Lineal")
plt.xlabel("Metros cuadrados")
plt.ylabel("Precio de la casa")
plt.legend()
plt.show()
```

```python
# ---------------------------------
# 5. Imprimir coeficientes encontrados
# ---------------------------------
print("Pendiente (coef_):", modelo.coef_[0])
print("Intercepción (intercept_):", modelo.intercept_)
```
