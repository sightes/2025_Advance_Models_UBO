```python
# 1. Importaciones necesarias
import numpy as np
from tensorflow.keras.datasets import mnist  # Para cargar MNIST
from sklearn.naive_bayes import GaussianNB   # Para Naive Bayes
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# 2. Cargar los datos de MNIST
#mnist.load_data() #devuelve dos tuplas: (x_train, y_train), (x_test, y_test)
# donde x contiene las imágenes y y las etiquetas (dígitos de 0 a 9).
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# 3. Preprocesamiento
# - Convertimos las imágenes de 28x28 a vectores de 784 dimensiones (flatten).
# - Convertimos de enteros a float para normalizar.
# - Escalamos al rango [0,1] dividiendo entre 255.

x_train = x_train.reshape(-1, 28*28).astype('float32') / 255.0
x_test  = x_test.reshape(-1, 28*28).astype('float32') / 255.0
```

```python
# 4. Definir y entrenar el modelo Naive Bayes
gnb = GaussianNB()
gnb.fit(x_train, y_train)
```

```python
# 5. Realizar predicciones sobre el conjunto de prueba
y_pred = gnb.predict(x_test)
```

```python
# 6. Evaluación del modelo
accuracy = accuracy_score(y_test, y_pred)
print(f"Exactitud en el conjunto de prueba: {accuracy:.4f}")
```

```python
# Reporte detallado de métricas
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred))
```

```python
# Matriz de confusión
print("Matriz de confusión:")
print(confusion_matrix(y_test, y_pred))
```
