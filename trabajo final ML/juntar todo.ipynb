```python
import pandas as pd
import pickle
import os
from glob import glob
import numpy as np

ruta_pkl = "./"  
archivos = ['results_new0105.pkl','results_new_ollama0105.pkl','results_new_exoglstm0105.pkl']
total_trials=pd.DataFrame([])
total_features=pd.DataFrame([])
for archivo in archivos:
    try:
        with open(archivo, "rb") as f:
            contenido = pickle.load(f)
            total_trials=pd.concat([total_trials,contenido['trials']])
            total_features=pd.concat([total_features,contenido['features']])
    except Exception as e:
        print(f"[ERROR] Falló al procesar {archivo}: {e}")
total_features=total_features.drop_duplicates()
results = {}
results['trials'] = total_trials
results['features']=total_features
with open('results.pkl', 'wb') as file:
    pickle.dump(results, file)
```
